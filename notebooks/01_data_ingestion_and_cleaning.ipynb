# Data Ingestion and Initial Cleaning

This notebook handles the loading of raw CSV data, merging of splits, and initial standardization.

```python
import sys
import os
import pandas as pd

# Add src to path
sys.path.append(os.path.abspath(os.path.join('..', 'src')))

from data_loader import load_all_data

# Set base directory
BASE_DIR = r"C:\Users\Deekshith J\OneDrive\Desktop\UDAI HACKATHON"

# Load data
data_dict = load_all_data(BASE_DIR)
```

## Save Processed Data
Saving the merged and standardized dataframes to parquet for faster access in downstream tasks.

```python
PROCESSED_DIR = os.path.join(BASE_DIR, 'data', 'processed')
os.makedirs(PROCESSED_DIR, exist_ok=True)

for key, df in data_dict.items():
    output_path = os.path.join(PROCESSED_DIR, f"{key}_processed.parquet")
    df.to_parquet(output_path, index=False)
    print(f"Saved {key} processed data to {output_path}")
```
