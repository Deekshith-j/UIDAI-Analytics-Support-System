# Data Cleaning Pipeline

This notebook verifies the cleaning logic implemented in `src/cleaning.py`.

```python
import sys
import os
import pandas as pd

sys.path.append(os.path.abspath(os.path.join('..', 'src')))

from data_loader import load_all_data
from cleaning import clean_dataset, clean_state_name

BASE_DIR = r"C:\Users\Deekshith J\OneDrive\Desktop\UDAI HACKATHON"
PROCESSED_DIR = os.path.join(BASE_DIR, 'data', 'processed')

# Load Pre-processed parquet files if they exist, else load raw (loading raw to be safe/consistent references)
# Actually, using load_all_data is safer to ensure we start from raw.
print("Loading Data...")
data = load_all_data(BASE_DIR)
```

```python
cleaned_data = {}

for key, df in data.items():
    print(f"\nCleaning {key} data...")
    clean_df = clean_dataset(df)
    cleaned_data[key] = clean_df
    
    print(f"Unique States in {key}: {sorted(clean_df['state'].unique())}")
```

```python
# Save Final Cleaned Data
for key, df in cleaned_data.items():
    output_path = os.path.join(PROCESSED_DIR, f"{key}_cleaned.parquet")
    df.to_parquet(output_path, index=False)
    print(f"Saved {key} cleaned data to {output_path}")
```
